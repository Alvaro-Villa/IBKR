```python
from ibkr.app import IBApp
from ibkr.contract import forex
from ibkr.order import market, bracket_order, BUY, SELL

import time

# Configurar períodos para MACD
short_period = 9
long_period = 12
signal_period = 9

# Inicializar la conexión con IBKR
app = IBApp("127.0.0.1", 7497, client_id=1, account="DUE310707")

# Configurar el contrato de divisas
cad = forex("USD/CAD")

try:
    while True:
        
        # Obtener datos históricos
        df = app.get_historical_data(99, cad, "120 S", "15 secs")
        
        # Calcular el True Range (TR) y Average True Range (ATR)
        df['tr'] = df[['high', 'low', 'close']].max(axis=1) - df[['high', 'low', 'close']].min(axis=1)
        atr = df['tr'].mean()

        # Calcular las medias móviles exponenciales (EMA) para el MACD
        df['ema_short'] = df['close'].ewm(span=short_period, adjust=False).mean()
        df['ema_long'] = df['close'].ewm(span=long_period, adjust=False).mean()

        # Calcular el MACD y la señal
        df['macd'] = df['ema_short'] - df['ema_long']
        df['signal'] = df['macd'].ewm(span=signal_period, adjust=False).mean()

        # Precio de cierre actual
        close = df.iloc[-1]['close']

        # Señal para abrir una posición Put (Sell)
        if (df.iloc[-1]['ema_short'] < df.iloc[-1]['ema_long']) and (df.iloc[-3]['ema_short'] > df.iloc[-3]['ema_long']):
            print("Señal de venta detectada. Abriendo posición Put...")

            orders = bracket_order(           
                        action=SELL,
                        quantity=25000,
                        take_profit_limit_price=close + (atr * 2),
                        stop_loss_price=close - (atr * 1.5)
                    )
                
            for o in orders:
                app.placeOrder(o.orderId, cad, o)
                print(o)
            
        elif (df.iloc[-1]['ema_short'] > df.iloc[-1]['ema_long']) and (df.iloc[-3]['ema_short'] < df.iloc[-3]['ema_long']):
            print("Señal de compra detectada. Abriendo posición Call...")

            orders = bracket_order(           
                        action=BUY,
                        quantity=25000,
                        take_profit_limit_price=close + (atr * 2),
                        stop_loss_price=close - (atr * 1.5)
            )

            
            for o in orders:
                app.placeOrder(o.orderId, cad, o)
                print(o)
        # Esperar antes de la próxima iteración
        time.sleep(20)

except KeyboardInterrupt:
    print("Ejecución interrumpida por el usuario.")
finally:
    app.disconnect()
    print("Conexión cerrada.")
```
