```python
import threading
import time

from ibkr.wrapper import IBWrapper
from ibkr.client import IBClient
from ibkr.contract import stock, future, option, forex


class IBApp(IBWrapper, IBClient):
    def __init__(self, ip, port, client_id):
        IBWrapper.__init__(self)
        IBClient.__init__(self, wrapper=self)

        self.connect(ip, port, client_id)

        thread = threading.Thread(target=self.run, daemon=True)
        thread.start()
        time.sleep(2)


if __name__ == "__main__":
    app = IBApp("127.0.0.1", 7497, client_id=5785)

    aapl = stock("AAPL", "SMART", "USD")
    gbl = future("GBL", "EUREX", "202403")
    pltr = option("PLTR", "BOX", "202503", 20, "C")
    gbp = forex("EUR/USD")
    data = app.get_market_data(
        request_id=99,
        contract=pltr,
    )

    time.sleep(30)
    app.disconnect()
```
